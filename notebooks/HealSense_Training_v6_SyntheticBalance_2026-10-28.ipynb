### What v6 changes

Only the **training split** may be rebalanced:

| `BALANCE_STRATEGY` | Meaning |
|---|---|
| `none` | Same-notebook baseline; real training windows only |
| `oversample` | Exact duplicate real minority-class training windows |
| `synthetic_augmentation` | New minority windows made by small physiological jitter/scaling of real training windows |

Safety rules enforced by the code:

1. Patients are split before any synthetic generation.
2. Synthetic windows are derived only from training patients.
3. Medians and the scaler are fitted on real training data only.
4. No synthetic window is added to validation or test.
5. Counts for real, synthetic, validation, and test data are printed and saved.

### Controlled comparison

Default is `synthetic_augmentation`. For a strict same-notebook baseline, after saving its outputs, restart the runtime and run again with `BALANCE_STRATEGY="none"`. Each strategy writes to a different output folder. Prefer validation macro-F1/critical recall for deciding whether the strategy helps; report the untouched test result once for the selected approach and disclose both runs in the thesis.

**Outputs:** `lstm_best.h5`, `scaler.pkl`, `label_encoder.pkl`, `imputer.pkl`, `model_meta.json`, and `evaluation_metrics.json`.

```python
DATA_ROOT = Path("/content/healsense_data" if IS_COLAB else "/kaggle/working/healsense_data")
DATA_ROOT.mkdir(parents=True, exist_ok=True)

REQUIRED = {"HR", "O2Sat", "Temp", "SBP", "DBP"}


def _valid_physionet_files(paths, sample=5):
    # True only if sampled files carry the 5 PhysioNet vital columns. Prevents
    # picking up a derived/pretty-printed CSV from someone else's notebook.
    if not paths:
        return False
    for fp in paths[:sample]:
        try:
            head = pd.read_csv(fp, sep="|" if fp.endswith(".psv") else ",", nrows=3)
        except Exception:
            return False
        if not REQUIRED.issubset({c.strip() for c in head.columns}):
            return False
    return True


def find_existing_psv():
    # Look for genuine PhysioNet patient files already mounted.
    # Order matters: a real dataset beats notebook outputs, and a file that
    # actually has the vital columns beats one that merely matches a filename.
    search_roots = []
    if IS_KAGGLE:
        print("Mounted inputs under /kaggle/input:")
        try:
            for d in sorted(os.listdir("/kaggle/input")):
                print(f"  - {d}")
        except Exception:
            pass
        search_roots.append("/kaggle/input/prediction-of-sepsis")   # known-good
        search_roots.append("/kaggle/input")
    if IS_COLAB and USE_DRIVE:
        try:
            from google.colab import drive
            if not os.path.ismount("/content/drive"):
                drive.mount("/content/drive")
            search_roots.append("/content/drive/MyDrive/healsense/data")
        except Exception as e:
            print(f"Drive mount skipped: {e}")
    search_roots.append(str(DATA_ROOT))

    seen = set()
    for root in search_roots:
        if not os.path.isdir(root) or root in seen:
            continue
        seen.add(root)
        for pattern in ("**/p*.psv", "**/p*.csv"):
            hits = sorted(glob.glob(os.path.join(root, pattern), recursive=True))
            if not hits:
                continue
            if _valid_physionet_files(hits):
                if "notebooks" in root:
                    print(f"WARNING: using NOTEBOOK OUTPUT under {root}")
                    print("         Attach the DATASET instead: salikhussaini49/prediction-of-sepsis")
                print(f"OK: {len(hits)} valid PhysioNet files under {root}")
                return hits
            print(f"   ({len(hits)} files under {root} lack HR/O2Sat/Temp/SBP/DBP - ignoring)")
    return []

files = find_existing_psv()

def _has_internet(timeout=6):
    try:
        import urllib.request
        urllib.request.urlopen("https://physionet-open.s3.amazonaws.com/", timeout=timeout)
        return True
    except Exception:
        return False


if not files and IS_KAGGLE:
    if _has_internet():
        print("NOTE: no dataset attached, but Internet is ON.")
        print("      Downloading from the PhysioNet S3 mirror (~5-8 min for 40k files).")
        print("      Faster next time: Add Input -> 'prediction-of-sepsis'")
        print("      (Salik Hussaini) -> Add, then re-run this cell.")
        print("")
    else:
        raise RuntimeError(
            "No PhysioNet data attached AND Internet is off. "
            "FIX (choose one): "
            "1) Right panel -> Add Input -> search 'prediction-of-sepsis' -> "
            "pick 'Prediction of Sepsis' by salikhussaini49 -> Add.  OR  "
            "2) Session options (right panel) -> Internet -> On, then re-run. "
            "Do NOT attach other people's NOTEBOOKS - they contain no raw data."
        )

if not files:
    print("No local copy found — downloading from the PhysioNet S3 mirror...")
    import concurrent.futures as cf, urllib.request

    BASE = "https://physionet-open.s3.amazonaws.com/challenge-2019/1.0.0/training"
    N_A, N_B = 20336, 20000
    # training_setB files are numbered p100001..p120000, NOT p000001.
    # Getting this wrong silently 404s the entire second half of the dataset.
    jobs = [("training_setA", f"p{k:06d}.psv") for k in range(1, N_A + 1)] + \
           [("training_setB", f"p{100000 + k:06d}.psv") for k in range(1, N_B + 1)]

    def fetch(job):
        subset, name = job
        dest = DATA_ROOT / subset / name
        if dest.exists() and dest.stat().st_size > 200:
            return str(dest)
        dest.parent.mkdir(parents=True, exist_ok=True)
        try:
            urllib.request.urlretrieve(f"{BASE}/{subset}/{name}", dest)
            if dest.stat().st_size < 200:   # 404 body
                dest.unlink(); return None
            return str(dest)
        except Exception:
            if dest.exists():
                dest.unlink()
            return None

    t0 = time.time()
    done = 0
    with cf.ThreadPoolExecutor(max_workers=48) as ex:
        for res in ex.map(fetch, jobs, chunksize=64):
            done += 1
            if done % 5000 == 0:
                print(f"  {done}/{len(jobs)} files | {time.time()-t0:.0f}s")
    files = sorted(str(p) for p in DATA_ROOT.rglob("*.psv") if p.stat().st_size > 200)

print(f"\nTotal patient files: {len(files)}")
if not files:
    raise RuntimeError("No data. On Kaggle add the dataset as input; on Colab re-run this cell.")
```

## 2 · Preprocess into windowed sequences

Labels are **future-window**: the worst severity in the `HORIZON` hours *after* the window ends. This is what removes the circularity in the original v1 pipeline.

```python
import numpy as np, pandas as pd

FEATURES = ["HR", "O2Sat", "Temp", "SBP", "DBP"]
CLASS_NAMES = ["normal", "warning", "critical"]
NORMAL, WARNING, CRITICAL = 0, 1, 2

# Mirror of backend/api/services/fusion_engine.py::evaluate_deterministic_rules
def severity_of_row(row):
    def val(c):
        v = row.get(c)
        try:
            v = float(v)
        except (TypeError, ValueError):
            return None
        return None if np.isnan(v) else v

    spo2, hr, temp, sbp, dbp = val("O2Sat"), val("HR"), val("Temp"), val("SBP"), val("DBP")
    if all(v is None for v in (spo2, hr, temp, sbp, dbp)):
        return None
    if (spo2 is not None and spo2 < 92) or \
       (hr is not None and (hr > 150 or hr < 40)) or \
       (sbp is not None and sbp >= 180) or (dbp is not None and dbp >= 120):
        return CRITICAL
    if (spo2 is not None and spo2 < 95) or \
       (hr is not None and (hr > 120 or hr < 50)) or \
       (temp is not None and (temp >= 38.0 or temp < 35.0)):
        return WARNING
    return NORMAL


def patient_windows(df, seq_len, horizon, min_valid=0.20):
    if len(df) < seq_len + 1:
        return None, None, None
    vals = df[FEATURES].to_numpy(dtype=np.float32)
    sev = np.full(len(df), -1, dtype=np.int8)
    for i in range(len(df)):
        s = severity_of_row(df.iloc[i])
        if s is not None:
            sev[i] = s

    X, y, idx = [], [], []
    for end in range(seq_len - 1, len(df)):
        fut = sev[end + 1: end + 1 + horizon]
        fut = fut[fut >= 0]
        if fut.size == 0:
            continue
        win = vals[end - seq_len + 1: end + 1]
        if np.mean(~np.isnan(win)) < min_valid:
            continue
        X.append(win); y.append(int(fut.max())); idx.append(end)
    if not X:
        return None, None, None
    return np.stack(X).astype(np.float32), np.asarray(y, np.int8), np.asarray(idx)


if N_PATIENTS_LIMIT:
    files = files[:N_PATIENTS_LIMIT]

# ── Robust loader ────────────────────────────────────────────────────────────
# Handles BOTH layouts that appear on Kaggle/Colab mirrors:
#   (a) one .psv file per patient          -> patient id = filename
#   (b) one combined CSV with a patient id column -> split by that column
# Without (b) a combined 1.55M-row CSV would be read as a SINGLE patient and
# sequences would bleed across patient boundaries.
PATIENT_ID_COLS = ["patient_id", "Patient_ID", "PatientID", "patient", "pid", "Patient"]


def load_patient_frames(paths):
    # Yield (patient_id, DataFrame) preserving original row order.
    for fp in paths:
        ext = fp.lower().rsplit(".", 1)[-1]
        sep = "|" if ext == "psv" else ","
        try:
            df = pd.read_csv(fp, sep=sep)
        except Exception:
            continue
        if any(c not in df.columns for c in FEATURES):
            continue
        idcol = next((c for c in PATIENT_ID_COLS if c in df.columns), None)
        if idcol is not None:
            for pid, g in df.groupby(idcol, sort=False):
                yield str(pid), g.reset_index(drop=True)
        else:
            yield Path(fp).stem, df


X_parts, y_parts, pid_parts, sep_parts = [], [], [], []
used = skipped = 0
t0 = time.time()

for pid, df in load_patient_frames(files):
    Xp, yp, idx = patient_windows(df, SEQ_LEN, HORIZON)
    if Xp is None:
        skipped += 1
        if N_PATIENTS_LIMIT and (used + skipped) >= N_PATIENTS_LIMIT:
            break
        continue
    used += 1
    X_parts.append(Xp); y_parts.append(yp)
    pid_parts.append(np.full(len(yp), pid, dtype=object))
    sep_parts.append(df["SepsisLabel"].to_numpy()[idx] if "SepsisLabel" in df.columns
                     else np.zeros(len(yp), np.int8))
    if used % 2000 == 0:
        print(f"  {used} patients | {time.time()-t0:.0f}s | "
              f"{sum(len(p) for p in y_parts)} windows")

X = np.concatenate(X_parts); y = np.concatenate(y_parts)
pids = np.concatenate(pid_parts); sepsis = np.concatenate(sep_parts).astype(np.int8)

print(f"\nDone in {time.time()-t0:.0f}s")
print(f"  patients used    : {used}  (skipped {skipped} — stay shorter than {SEQ_LEN}h)")
print(f"  windows          : {len(y)}")
print(f"  X shape          : {X.shape}")
for i, n in enumerate(CLASS_NAMES):
    print(f"    {n:9s}: {(y==i).sum():>8} ({(y==i).mean()*100:5.2f}%)")
print("  missingness      : " + ", ".join(
    f"{f} {np.isnan(X[:,:,i]).mean()*100:.1f}%" for i, f in enumerate(FEATURES)))
```

## 3 · Patient-wise split + imputation + scaling

Split by **patient**, never by window. The optional synthetic balancing happens **after** this split, uses training windows only, and never alters validation or test data. Medians and the scaler are fitted on real training windows only.


```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

# Patient-wise split, stratified by whether the patient ever had a critical window
crit_by_pid = pd.Series(y).groupby(pids).max()
pid_train, pid_tmp = train_test_split(
    crit_by_pid.index.values, test_size=0.30, random_state=SEED,
    stratify=crit_by_pid.values)
pid_val, pid_test = train_test_split(
    pid_tmp, test_size=0.50, random_state=SEED,
    stratify=crit_by_pid.loc[pid_tmp].values)

tr = np.isin(pids, pid_train); va = np.isin(pids, pid_val); te = np.isin(pids, pid_test)
X_tr, y_tr = X[tr], y[tr]
X_va, y_va = X[va], y[va]
X_te, y_te = X[te], y[te]
print(f"train {len(y_tr):>7} windows / {len(pid_train)} patients")
print(f"val   {len(y_va):>7} windows / {len(pid_val)} patients")
print(f"test  {len(y_te):>7} windows / {len(pid_test)} patients")

# Keep explicit real-training references. All balancing below starts from these
# arrays; X_va/y_va and X_te/y_te are never used as synthetic sources.
X_tr_real, y_tr_real = X_tr, y_tr

def class_counts(labels):
    return {CLASS_NAMES[i]: int((labels == i).sum()) for i in range(3)}

def print_class_counts(title, labels):
    counts = class_counts(labels)
    total = max(1, sum(counts.values()))
    print(title)
    for name, count in counts.items():
        print(f"  {name:9s}: {count:>8,} ({100*count/total:5.2f}%)")
    return counts

REAL_TRAIN_COUNTS = print_class_counts("\nReal TRAIN windows (before balancing):", y_tr_real)
VAL_COUNTS = print_class_counts("Real VALIDATION windows (never augmented):", y_va)
TEST_COUNTS = print_class_counts("Real TEST windows (never augmented):", y_te)

# ── Medians from TRAIN ONLY (no leakage) ──
FALLBACKS = dict(zip(FEATURES, [75.0, 98.0, 37.0, 120.0, 80.0]))
flat = X_tr.reshape(-1, len(FEATURES))
medians = {}
print("\nTraining-split medians:")
for i, f in enumerate(FEATURES):
    col = flat[:, i]
    medians[f] = FALLBACKS[f] if np.all(np.isnan(col)) else float(np.nanmedian(col))
    print(f"  {f:6s} {medians[f]:7.2f}  (missing {np.mean(np.isnan(col))*100:4.1f}%)")


# ── Optional TRAINING-SPLIT-ONLY balancing ───────────────────────────────────
# "Synthetic" here means physiologically bounded augmentation of real training
# sequences, not invented patients and not synthetic validation/test data.
PHYS_BOUNDS = {
    "HR": (25.0, 220.0),
    "O2Sat": (50.0, 100.0),
    "Temp": (30.0, 43.0),
    "SBP": (50.0, 260.0),
    "DBP": (25.0, 180.0),
}
SYNTHETIC_COUNTS = {name: 0 for name in CLASS_NAMES}
X_syn_raw = np.empty((0, SEQ_LEN, len(FEATURES)), dtype=np.float32)
y_syn = np.empty((0,), dtype=np.int8)

if BALANCE_STRATEGY not in {"none", "oversample", "synthetic_augmentation"}:
    raise ValueError(f"Unknown BALANCE_STRATEGY={BALANCE_STRATEGY!r}")

if BALANCE_STRATEGY != "none":
    import warnings
    rng = np.random.default_rng(SEED + 20261028)
    counts_arr = np.array([(y_tr_real == i).sum() for i in range(3)], dtype=int)
    majority_count = int(counts_arr.max())
    target_count = int(round(majority_count * SYNTHETIC_TARGET_RATIO))
    syn_parts, syn_labels = [], []

    for cls in (WARNING, CRITICAL):
        src_idx = np.flatnonzero(y_tr_real == cls)
        n_new = min(
            max(0, target_count - int(counts_arr[cls])),
            int(len(src_idx) * SYNTHETIC_MAX_PER_ORIGINAL),
        )
        if n_new == 0:
            continue

        # Tile then shuffle source indices: guarantees no original window is
        # reused more than SYNTHETIC_MAX_PER_ORIGINAL times.
        reuse_pool = np.tile(src_idx, SYNTHETIC_MAX_PER_ORIGINAL)
        rng.shuffle(reuse_pool)
        chosen = reuse_pool[:n_new]
        aug = X_tr_real[chosen].astype(np.float32, copy=True)

        if BALANCE_STRATEGY == "synthetic_augmentation":
            observed = ~np.isnan(aug)
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", category=RuntimeWarning)
                means = np.nanmean(aug, axis=1, keepdims=True)
            means = np.where(np.isnan(means), 0.0, means)
            scale = rng.uniform(
                SYNTHETIC_SCALE_RANGE[0], SYNTHETIC_SCALE_RANGE[1],
                size=(n_new, 1, len(FEATURES)),
            ).astype(np.float32)
            aug = means + (aug - means) * scale

            sigma = np.array([SYNTHETIC_JITTER[f] for f in FEATURES], dtype=np.float32)
            noise = rng.normal(0.0, 1.0, size=aug.shape).astype(np.float32) * sigma
            aug = np.where(observed, aug + noise, aug)

            for j, feature in enumerate(FEATURES):
                lo, hi = PHYS_BOUNDS[feature]
                aug[:, :, j] = np.where(
                    observed[:, :, j],
                    np.clip(aug[:, :, j], lo, hi),
                    aug[:, :, j],
                )

            if SYNTHETIC_EXTRA_MISSING_RATE > 0:
                extra_missing = observed & (
                    rng.random(aug.shape) < SYNTHETIC_EXTRA_MISSING_RATE
                )
                aug[extra_missing] = np.nan
            aug[~observed] = np.nan

        syn_parts.append(aug.astype(np.float32, copy=False))
        syn_labels.append(np.full(n_new, cls, dtype=np.int8))
        SYNTHETIC_COUNTS[CLASS_NAMES[cls]] = int(n_new)

    if syn_parts:
        X_syn_raw = np.concatenate(syn_parts)
        y_syn = np.concatenate(syn_labels)

print(f"\nBalance strategy: {BALANCE_STRATEGY}")
print(f"Generated/duplicated TRAINING windows: {len(y_syn):,}")
for name, count in SYNTHETIC_COUNTS.items():
    print(f"  {name:9s}: {count:>8,}")


def featurize(Xin, medians, scaler=None, fit=False):
    """Impute -> append 5 missing-indicators -> scale the 5 raw channels only."""
    Xw = Xin.astype(np.float32, copy=True)
    ind = np.isnan(Xw).astype(np.float32)          # 1.0 = missing
    for i, f in enumerate(FEATURES):
        col = Xw[:, :, i]
        col[np.isnan(col)] = medians[f]
    n, t, _ = Xw.shape
    flat = Xw.reshape(-1, len(FEATURES))
    flat = (scaler.fit_transform(flat) if fit else scaler.transform(flat))
    Xw = flat.reshape(n, t, len(FEATURES))
    return np.concatenate([Xw, ind], axis=2).astype(np.float32)   # (n, t, 10)


scaler = StandardScaler()
# Fit scaler on REAL training windows only. Synthetic windows must not move
# the scaler or medians.
X_tr_f = featurize(X_tr_real, medians, scaler, fit=True)
y_tr = y_tr_real

if len(y_syn):
    X_syn_f = featurize(X_syn_raw, medians, scaler)
    X_tr_f = np.concatenate([X_tr_f, X_syn_f])
    y_tr = np.concatenate([y_tr_real, y_syn])
    del X_syn_f

# Shuffle final training order once; Keras also shuffles each epoch.
shuffle_idx = np.random.default_rng(SEED).permutation(len(y_tr))
X_tr_f, y_tr = X_tr_f[shuffle_idx], y_tr[shuffle_idx]

X_va_f = featurize(X_va, medians, scaler)
X_te_f = featurize(X_te, medians, scaler)

FINAL_TRAIN_COUNTS = print_class_counts("\nFinal TRAIN windows (real + allowed train-only additions):", y_tr)
assert len(y_va) == sum(VAL_COUNTS.values()), "Validation set changed unexpectedly"
assert len(y_te) == sum(TEST_COUNTS.values()), "Test set changed unexpectedly"

# Labels are ALREADY 0=normal, 1=warning, 2=critical.
# Bind a LabelEncoder with that exact order for backend compatibility.
# Fitting on the names would sort them alphabetically (critical, normal,
# warning) and silently remap every class — a real bug caught in testing.
# Free the raw (unfeaturized) copies before training — on the full 40k-patient
# cohort the array is ~800 MB and duplicating it can exhaust notebook RAM.
del X, X_tr, X_va, X_tr_real, X_syn_raw   # keep X_te for the rules-baseline ablation
import gc; gc.collect()

le = LabelEncoder()
le.classes_ = np.asarray(CLASS_NAMES)
assert list(le.inverse_transform([0, 1, 2])) == CLASS_NAMES, "class order mismatch"
y_tr_e, y_va_e, y_te_e = y_tr, y_va, y_te
y_tr_c = tf.keras.utils.to_categorical(y_tr_e, 3)
y_va_c = tf.keras.utils.to_categorical(y_va_e, 3)
y_te_c = tf.keras.utils.to_categorical(y_te_e, 3)

cw = compute_class_weight("balanced", classes=np.unique(y_tr_e), y=y_tr_e)
class_weight = dict(enumerate(cw))
print(f"\nFinal input shape: {X_tr_f.shape}")
print(f"Class weights    : {dict(zip(CLASS_NAMES, np.round(cw,3)))}")
```

## 4 · Build the model

**On "maximum parameters":** the model has 5 input signals and ~600k windows. Going much past ~150k parameters mostly buys you **overfitting to the rule engine's own labels**, not accuracy — because the labels are a deterministic function of 5 numbers. The cell below prints the train/val gap after training; if it exceeds ~0.05, go *smaller*, not larger. The real lever for your grade is the ablation in step 6, not parameter count.

```python
import tensorflow as tf
from tensorflow.keras import layers, Sequential
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau

tf.random.set_seed(SEED); np.random.seed(SEED)

n_feat = X_tr_f.shape[2]
model = Sequential([
    layers.Input(shape=(SEQ_LEN, n_feat)),
    layers.LSTM(CFG["lstm"][0], return_sequences=True),
    layers.Dropout(CFG["dr"][0]),
    layers.LSTM(CFG["lstm"][1]),
    layers.Dropout(CFG["dr"][1]),
    layers.Dense(CFG["dense"], activation="relu"),
    layers.Dropout(CFG["dr"][2]),
    layers.Dense(3, activation="softmax"),
])
model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
              loss="categorical_crossentropy", metrics=["accuracy"])
model.summary()
print(f"\nTotal parameters: {model.count_params():,}")
```

## 6 · Evaluate — including the ablation that answers *"what does the AI add?"*

```python
from sklearn.metrics import (
    accuracy_score, average_precision_score, balanced_accuracy_score,
    classification_report, confusion_matrix, f1_score,
    precision_recall_fscore_support, roc_auc_score,
)
from sklearn.preprocessing import label_binarize

probs = model.predict(X_te_f, batch_size=512, verbose=0)
pred = probs.argmax(1)
model_class = np.array(CLASS_NAMES)[pred]

print("=" * 62)
print("LSTM vs FUTURE-WINDOW GROUND TRUTH")
print("=" * 62)
print(classification_report(y_te, pred, target_names=CLASS_NAMES, digits=3, zero_division=0))
print(confusion_matrix(y_te, pred))

# ── Rules-only baseline on the SAME windows ──
# "Current state" severity = what the deterministic engine would say right now.
def current_severity(Xraw):
    out = np.zeros(len(Xraw), np.int8)
    hr, spo2, temp, sbp, dbp = (Xraw[:, -1, i] for i in range(5))
    with np.errstate(invalid="ignore"):
        crit = ((spo2 < 92) | (hr > 150) | (hr < 40) | (sbp >= 180) | (dbp >= 120))
        warn = ((spo2 < 95) | (hr > 120) | (hr < 50) | (temp >= 38) | (temp < 35))
    out[np.nan_to_num(warn, nan=False).astype(bool)] = WARNING
    out[np.nan_to_num(crit, nan=False).astype(bool)] = CRITICAL
    return out

rules_now = current_severity(X_te)

print("")
print("ABLATION — predicted vs FUTURE labels (higher F1 = better early warning)")
print("=" * 62)
rows = []
for name, p in [("Rules only (now)", rules_now), ("LSTM only", pred)]:
    rows.append((name,
                 f1_score(y_te, p, average="macro", zero_division=0),
                 f1_score(y_te, p, average=None, zero_division=0)))

# (a) EXACT mirror of backend fuse_risk(): weighted 0.6*det + 0.4*ml, critical capped.
DET_SCORE = np.array([0.0, 0.5, 1.0])      # normal, warning, critical
ML_SCORE  = np.array([0.0, 0.5, 1.0])
weighted = 0.6 * DET_SCORE[rules_now] + 0.4 * ML_SCORE[pred]
fuse_w = np.where(weighted >= 0.7, CRITICAL, np.where(weighted >= 0.4, WARNING, NORMAL))
fuse_w = np.where(rules_now == CRITICAL, CRITICAL, fuse_w)                      # absolute override
fuse_w = np.where((fuse_w == CRITICAL) & (rules_now != CRITICAL), WARNING, fuse_w)  # safety cap

# (b) Safety-first OR: EITHER layer may raise an alarm, but only the rules may
# declare CRITICAL. Keeps the ML's warning signal instead of discarding it.
fuse_or = np.maximum(rules_now, pred)
fuse_or = np.where((fuse_or == CRITICAL) & (rules_now != CRITICAL), WARNING, fuse_or)

for label, p in [("Fusion A: weighted 0.6/0.4 (backend)", fuse_w),
                 ("Fusion B: safety-first OR", fuse_or)]:
    rows.append((label,
                 f1_score(y_te, p, average="macro", zero_division=0),
                 f1_score(y_te, p, average=None, zero_division=0)))

print(f"{'Method':34s} {'macro-F1':>9s}  " + "  ".join(f"{n[:8]:>9s}" for n in CLASS_NAMES))
for name, mf1, per in rows:
    print(f"{name:34s} {mf1:9.3f}  " + "  ".join(f"{v:9.3f}" for v in per))

# ── Fusion weight sweep: is 0.6/0.4 actually the best split? ──
# The backend uses DETERMINISTIC_WEIGHT=0.6 / ML_WEIGHT=0.4 (a design choice,
# never tuned). This reports macro-F1 for every split so the report can justify
# the weights with evidence instead of asserting them.
print("")
print("FUSION WEIGHT SWEEP  (det weight -> ml weight)")
print("=" * 62)
best_w, best_f1 = 0.6, -1.0
for w in np.arange(0.0, 1.01, 0.1):
    sc = w * DET_SCORE[rules_now] + (1 - w) * ML_SCORE[pred]
    fz = np.where(sc >= 0.7, CRITICAL, np.where(sc >= 0.4, WARNING, NORMAL))
    fz = np.where(rules_now == CRITICAL, CRITICAL, fz)
    fz = np.where((fz == CRITICAL) & (rules_now != CRITICAL), WARNING, fz)
    f1w = f1_score(y_te, fz, average="macro", zero_division=0)
    tag = "  <-- backend default" if abs(w - 0.6) < 1e-9 else ""
    marker = " *best*" if f1w > best_f1 else ""
    if f1w > best_f1:
        best_f1, best_w = f1w, w
    print(f"  det {w:.1f} / ml {1-w:.1f}   macro-F1 {f1w:.3f}{marker}{tag}")
print("")
print(f"Best split: det {best_w:.1f} / ml {1-best_w:.1f}  ->  {best_f1:.3f}")
if abs(best_w - 0.6) > 0.05:
    print("NOTE: the backend default (0.6) is NOT optimal on this data.")
    print(f"      Consider DETERMINISTIC_WEIGHT={best_w:.1f} / ML_WEIGHT={1-best_w:.1f}")
    print("      in backend/api/config.py, and quote this sweep in the report.")

# ── The headline metric: early warning capture ──
pre_det = (y_te == CRITICAL) & (rules_now == NORMAL)
if pre_det.sum() > 0:
    caughtted = ((model_class == "critical") | (model_class == "warning"))[pre_det].mean()
    print(f"\nEARLY-WARNING CAPTURE (the number for your report)")
    print(f"  windows heading toward a critical state within {HORIZON}h, where the")
    print(f"  rules engine currently reports NORMAL : {pre_det.sum():,} windows")
    print(f"  fraction the LSTM already flags as warning/critical: {caughtted*100:.1f}%")
    print(f"  fraction the rules engine flags                     : 0.0%  (by definition)")

# ── Complete metric pack: the numbers needed for the final report ────────────
# Validation is for comparing none / oversample / synthetic. Test remains a
# real, natural-prevalence hold-out and contains zero synthetic windows.

def metric_pack(y_true, probabilities):
    y_hat = probabilities.argmax(1)
    precision, recall, f1, support = precision_recall_fscore_support(
        y_true, y_hat, labels=[0, 1, 2], zero_division=0
    )
    pack = {
        "accuracy": float(accuracy_score(y_true, y_hat)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true, y_hat)),
        "macro_f1": float(f1_score(y_true, y_hat, average="macro", zero_division=0)),
        "weighted_f1": float(f1_score(y_true, y_hat, average="weighted", zero_division=0)),
        "per_class": {
            CLASS_NAMES[i]: {
                "precision": float(precision[i]),
                "recall": float(recall[i]),
                "f1": float(f1[i]),
                "support": int(support[i]),
            }
            for i in range(3)
        },
        "confusion_matrix": confusion_matrix(y_true, y_hat, labels=[0, 1, 2]).tolist(),
    }
    try:
        pack["auroc_macro_ovr"] = float(
            roc_auc_score(y_true, probabilities, multi_class="ovr", average="macro")
        )
        y_bin = label_binarize(y_true, classes=[0, 1, 2])
        pack["auprc_per_class"] = {
            CLASS_NAMES[i]: float(average_precision_score(y_bin[:, i], probabilities[:, i]))
            for i in range(3)
        }
    except Exception as exc:  # keep the core metrics even if a split lacks a class
        pack["ranking_metric_error"] = f"{type(exc).__name__}: {exc}"
    return pack

val_probs = model.predict(X_va_f, batch_size=512, verbose=0)
EVAL_METRICS = {
    "notebook_build": NOTEBOOK_VERSION,
    "run_name": RUN_NAME,
    "balance_strategy": BALANCE_STRATEGY,
    "synthetic_target_ratio": float(SYNTHETIC_TARGET_RATIO),
    "class_counts": {
        "real_train": REAL_TRAIN_COUNTS,
        "synthetic_or_duplicated_train": SYNTHETIC_COUNTS,
        "final_train": FINAL_TRAIN_COUNTS,
        "validation_real_only": VAL_COUNTS,
        "test_real_only": TEST_COUNTS,
    },
    "validation": metric_pack(y_va, val_probs),
    "test": metric_pack(y_te, probs),
}

print("\n" + "=" * 62)
print("FINAL METRIC SUMMARY — copy/save these values exactly")
print("=" * 62)
for split in ("validation", "test"):
    m = EVAL_METRICS[split]
    print(
        f"{split:10s} accuracy={m['accuracy']:.4f}  "
        f"balanced_accuracy={m['balanced_accuracy']:.4f}  "
        f"macro_F1={m['macro_f1']:.4f}  weighted_F1={m['weighted_f1']:.4f}"
    )
    for name in CLASS_NAMES:
        c = m["per_class"][name]
        print(
            f"  {name:9s} P={c['precision']:.3f} R={c['recall']:.3f} "
            f"F1={c['f1']:.3f} support={c['support']:,}"
        )

with open(OUT_DIR / "evaluation_metrics.json", "w") as f:
    json.dump(EVAL_METRICS, f, indent=2)
print(f"\nSaved metrics → {OUT_DIR / 'evaluation_metrics.json'}")

```
